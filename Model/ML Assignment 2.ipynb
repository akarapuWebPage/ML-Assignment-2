```python

# ============================================================
# BANK MARKETING CLASSIFICATION
# Machine Learning Assignment
#
# Dataset: UCI Bank Marketing - bank-full.csv
# Classification: Binary (yes / no)
#
# Methodology:
#   1. Original, imbalanced dataset
#   2. Exclude duration to avoid data leakage
#   3. Stratified 80:20 train/test split
#   4. Five classification algorithms
#   5. Default prediction thresholds
#   6. Six evaluation metrics
#   7. Training vs testing accuracy gap
#
# ============================================================


# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import os
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

# Create an output folder for results and plots.
OUTPUT_DIR = "."
os.makedirs(OUTPUT_DIR, exist_ok=True)

sns.set_theme(style="whitegrid")


# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

# Keep bank-full.csv in the same directory as this script.
DATASET_PATH = "bank-full.csv"

# The UCI Bank Marketing full dataset uses a semicolon delimiter.
# Load Bank Marketing dataset using Python parsing engine
df = pd.read_csv(
    DATASET_PATH,
    sep=";",
    engine="python"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
print("\n========== DATASET OVERVIEW ==========")

print("\nFirst five records:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())


# ------------------------------------------------------------
# 3. TARGET VARIABLE DISTRIBUTION
# ------------------------------------------------------------

print("\n========== TARGET DISTRIBUTION ==========")

print(df["y"].value_counts())

print("\nTarget distribution in percentage:")
print(df["y"].value_counts(normalize=True) * 100)

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="y",
    order=["no", "yes"]
)

plt.title("Bank Marketing Target Distribution")
plt.xlabel("Subscription")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/target_distribution.png",
    dpi=300
)

plt.show()


# ------------------------------------------------------------
# 4. FEATURE SELECTION
# ------------------------------------------------------------

# duration is excluded because it represents the length of
# the phone call. It is only known after the call has ended.
# Using it would introduce data leakage for our prediction goal.

df_model = df.drop(columns=["duration"])

# Separate predictors and target.
X = df_model.drop(columns=["y"])

# Convert target values into numerical labels.
# no  = 0
# yes = 1
y = df_model["y"].map({
    "no": 0,
    "yes": 1
})

print("\n========== FEATURE INFORMATION ==========")

print("\nInput feature count:", X.shape[1])
print("Target variable: y")

print("\nInput features:")
print(X.columns.tolist())

# Identify numerical and categorical features.
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

print("\nNumber of numerical features:", len(numeric_features))
print("Number of categorical features:", len(categorical_features))


# ------------------------------------------------------------
# 5. EXPLORATORY DATA ANALYSIS
# ------------------------------------------------------------

# 5.1 Distribution of numerical features.

X[numeric_features].hist(
    figsize=(14, 10),
    bins=30,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/numerical_distributions.png",
    dpi=300
)

plt.show()


# 5.2 Correlation heatmap for numerical features.

plt.figure(figsize=(10, 7))

correlation_matrix = df[numeric_features].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap - Numerical Features")
plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/correlation_heatmap.png",
    dpi=300
)

plt.show()


# 5.3 Subscription by job type.

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="job",
    hue="y",
    order=df["job"].value_counts().index
)

plt.title("Customer Subscription by Job Type")
plt.xlabel("Job Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/subscription_by_job.png",
    dpi=300
)

plt.show()


# 5.4 Subscription by education.

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="education",
    hue="y"
)

plt.title("Customer Subscription by Education")
plt.xlabel("Education")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/subscription_by_education.png",
    dpi=300
)

plt.show()


# ------------------------------------------------------------
# 6. TRAIN / TEST SPLIT
# ------------------------------------------------------------

# Use a single stratified 80:20 split.
#
# Stratification preserves the original yes/no class
# proportions in both training and testing datasets.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("\n========== TRAIN / TEST SPLIT ==========")

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())

# Save the held-out test data for Streamlit
test_data = X_test.copy()

test_data["y"] = y_test.map({
    0: "no",
    1: "yes"
})

test_data.to_csv("../test_data.csv", index=False)
print("\n")
print("Test data saved successfully!")

# ------------------------------------------------------------
# 7. DATA PREPROCESSING
# ------------------------------------------------------------

# Numerical features:
# StandardScaler brings numerical values to a comparable scale.
#
# Categorical features:
# OneHotEncoder converts categories into numerical columns.
#
# The preprocessing is fitted only on training data through
# the Pipeline. The test data is transformed using the same
# fitted preprocessing steps.

numeric_transformer = StandardScaler()

categorical_transformer = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ------------------------------------------------------------
# 8. DEFINE CLASSIFICATION MODELS
# ------------------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1500,
        random_state=RANDOM_STATE
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=RANDOM_STATE,
        min_samples_leaf=5
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5,
        n_jobs=-1
    ),

    "Gaussian Naive Bayes": GaussianNB(),

    "SVM": LinearSVC(
        random_state=RANDOM_STATE,
        max_iter=10000
    )
}


# ------------------------------------------------------------
# 9. TRAIN MODELS AND EVALUATE PERFORMANCE
# ------------------------------------------------------------

results = []

trained_models = {}

confusion_matrices = {}

for model_name, classifier in models.items():

    print("\n" + "=" * 65)
    print("MODEL:", model_name)
    print("=" * 65)

    # Create a complete pipeline containing preprocessing
    # followed by the classification algorithm.

    pipeline = Pipeline(
        steps=[
            ("preprocessing", preprocessor),
            ("classifier", classifier)
        ]
    )

    # Train using the original training dataset.
    pipeline.fit(X_train, y_train)

    # Store the trained pipeline.
    trained_models[model_name] = pipeline

    # Predictions on training and testing datasets.
    y_train_pred = pipeline.predict(X_train)
    y_test_pred = pipeline.predict(X_test)

    # --------------------------------------------------------
    # 9.1 Calculate training and testing accuracy
    # --------------------------------------------------------

    train_accuracy = accuracy_score(
        y_train,
        y_train_pred
    )

    test_accuracy = accuracy_score(
        y_test,
        y_test_pred
    )

    # Accuracy gap in percentage points.
    accuracy_gap = (
        train_accuracy - test_accuracy
    ) * 100

    # --------------------------------------------------------
    # 9.2 Calculate AUC
    # --------------------------------------------------------

    # Models with predict_proba provide positive-class
    # probabilities.
    #
    # LinearSVC does not provide predict_proba by default.
    # Therefore, its decision_function scores are used for AUC.

    if hasattr(pipeline, "predict_proba"):

        y_test_score = pipeline.predict_proba(
            X_test
        )[:, 1]

    else:

        y_test_score = pipeline.decision_function(
            X_test
        )

    auc = roc_auc_score(
        y_test,
        y_test_score
    )

    # --------------------------------------------------------
    # 9.3 Calculate remaining evaluation metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        y_test_pred
    )

    precision = precision_score(
        y_test,
        y_test_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_test_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_test_pred,
        zero_division=0
    )

    mcc = matthews_corrcoef(
        y_test,
        y_test_pred
    )

    # --------------------------------------------------------
    # 9.4 Store model results
    # --------------------------------------------------------

    results.append({
        "Model": model_name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": accuracy,
        "Accuracy Gap (pp)": accuracy_gap,
        "AUC": auc,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "MCC": mcc
    })

        # --------------------------------------------------------
    # 9.5 Generate confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_test,
        y_test_pred
    )

    confusion_matrices[model_name] = cm

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_test_pred,
            target_names=["No", "Yes"],
            zero_division=0
        )
    )

    # Create and save confusion matrix plot.
    cm_display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["No", "Yes"]
    )

    cm_display.plot(
        cmap="Blues",
        values_format="d"
    )

    plt.title(f"Confusion Matrix - {model_name}")
    plt.tight_layout()

    safe_name = model_name.lower().replace(" ", "_")

    plt.savefig(
        f"{OUTPUT_DIR}/confusion_matrix_{safe_name}.png",
        dpi=300
    )

    plt.show()
    plt.close()

# ------------------------------------------------------------
# 10. CONSOLIDATED MODEL COMPARISON
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

# Display metrics with four decimal places.
print("\n========== FINAL MODEL COMPARISON ==========")

print(
    results_df.round(4).to_string(index=False)
)

# Save results to CSV.
results_df.to_csv(
    f"{OUTPUT_DIR}/model_comparison.csv",
    index=False
)


# ------------------------------------------------------------
# 11. VISUALIZE MODEL PERFORMANCE
# ------------------------------------------------------------

metrics_to_plot = [
    "Testing Accuracy",
    "AUC",
    "Precision",
    "Recall",
    "F1 Score",
    "MCC"
]

for metric in metrics_to_plot:

    plt.figure(figsize=(10, 5))

    sns.barplot(
        data=results_df,
        x="Model",
        y=metric
    )

    plt.title(f"Model Comparison - {metric}")
    plt.xlabel("Classification Model")
    plt.ylabel(metric)
    plt.xticks(rotation=20)

    plt.tight_layout()

    safe_metric = metric.lower().replace(" ", "_")

    plt.savefig(
        f"{OUTPUT_DIR}/{safe_metric}_comparison.png",
        dpi=300
    )

    plt.show()


# ------------------------------------------------------------
# 12. ACCURACY GAP COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.barplot(
    data=results_df,
    x="Model",
    y="Accuracy Gap (pp)"
)

plt.axhline(
    y=0,
    linestyle="--",
    color="black"
)

plt.title("Training vs Testing Accuracy Gap")
plt.xlabel("Classification Model")
plt.ylabel("Accuracy Gap (percentage points)")
plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/accuracy_gap_comparison.png",
    dpi=300
)

plt.show()


# ------------------------------------------------------------
# 13. IDENTIFY BEST MODEL FOR EACH METRIC
# ------------------------------------------------------------

print("\n========== BEST MODEL BY METRIC ==========")

for metric in [
    "Testing Accuracy",
    "AUC",
    "Precision",
    "Recall",
    "F1 Score",
    "MCC"
]:

    best_index = results_df[metric].idxmax()

    best_model = results_df.loc[
        best_index,
        "Model"
    ]

    best_value = results_df.loc[
        best_index,
        metric
    ]

    print(
        f"{metric}: {best_model} ({best_value:.4f})"
    )


# ------------------------------------------------------------
# 14. SAVE TRAINED MODELS
# ------------------------------------------------------------

# Save each complete pipeline.
# This includes both preprocessing and the classifier.

for model_name, pipeline in trained_models.items():

    safe_name = model_name.lower().replace(" ", "_")

    joblib.dump(
        pipeline,
        f"{OUTPUT_DIR}/{safe_name}.joblib"
    )


# ------------------------------------------------------------
# 15. SAVE FEATURE INFORMATION
# ------------------------------------------------------------

feature_information = {
    "input_features": X.columns.tolist(),
    "numerical_features": numeric_features,
    "categorical_features": categorical_features,
    "target": "y",
    "target_mapping": {
        "no": 0,
        "yes": 1
    }
}

joblib.dump(
    feature_information,
    f"{OUTPUT_DIR}/feature_information.joblib"
)


# ------------------------------------------------------------
# 16. FINAL SUMMARY
# ------------------------------------------------------------

print("\n========== EXPERIMENT COMPLETED ==========")

print("Dataset records:", len(df))
print("Input features:", X.shape[1])
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Models trained:", len(trained_models))

print("\nFiles saved in:", OUTPUT_DIR)

print("\nGenerated files include:")
print("- Model comparison CSV")
print("- Confusion matrix plots")
print("- Metric comparison plots")
print("- Accuracy gap comparison")
print("- Five trained model pipelines")
print("- Feature information")

print("\nAll experiments completed successfully.")
```
